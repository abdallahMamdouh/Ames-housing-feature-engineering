## Merage data

```python
emp = pd.DataFrame(
    {
        "emp_id": [1,2,3,4],
        "names": ["Ahmed", "Mohamed", "Alaa", "Aya"],
        "d_id": [100, 101, 105, 104]
    }
)

department = pd.DataFrame(
    {
        "d_id": [100, 101, 200],
        "d_name": ["IT", "HR", "CS"]
    }
)
```

```python
inner_join = pd.merge(emp, department, on="d_id", how="inner")
inner_join
```

```python
left_join = pd.merge(emp, department, on="d_id", how="left")
left_join
```

```python
right_join = pd.merge(emp, department, on="d_id", how="right")
right_join
```

```python
outer_join = pd.merge(emp, department, on="d_id", how="outer")
outer_join
```

## Handle missing values

```python
df = pd.DataFrame(
    {
        "age": [20,25,np.nan, 30,35]
    }
)
```

```python
df
```

```python
# Mean Imputation
df['age_mean'] = df['age'].fillna(df['age'].mean())
df
```

```python
# Median Imputation
df['age_median'] = df['age'].fillna(df['age'].median())
df
```

```python
# FFill
df['age_ffil'] = df['age'].fillna(method='ffill')
```

```python
df
```

```python
# BFill
df['age_bfil'] = df['age'].fillna(method='bfill')
df
```

```python
def check_nan(df):
    precent = df.isnull().sum()/len(df)*100
    precent = precent[precent>0].sort_values()
    return precent
```

```python
values = check_nan(df)
values
```
